```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize']=[25,3]
img=cv2.imread('a1.png',cv2.IMREAD_COLOR)
imgRGB=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

r=img[:,:,0]
g=img[:,:,1]
b=img[:,:,2]

r_img = np.zeros_like(imgRGB)
r_img[:,:,0] = r

g_img = np.zeros_like(imgRGB)
g_img[:,:,1] = g

b_img = np.zeros_like(imgRGB)
b_img[:,:,2] = b

plt.figure(figsize=(25,5))

plt.subplot(141)
plt.title('imgane original')
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(142)
plt.title('imgane original')
plt.imshow(r_img)
plt.axis('off')

plt.subplot(143)
plt.title('imgane original')
plt.imshow(g_img)
plt.axis('off')

plt.subplot(144)
plt.title('imgane original')
plt.imshow(b_img)
plt.axis('off')

plt.show
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
img=cv2.imread('morado.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

fig, axs = plt.subplots(1, 4, figsize=(20,5))

axs[0].imshow(img_rgb)
axs[0].set_title('imagenhard')

fig_img=axs[1].imshow(img_rgb[...,0],cmap='gray')
fig.colorbar(fig_img,ax=axs[1],shrink=0.5)
axs[1].set_title('canal rojo')

fig_img=axs[2].imshow(img_rgb[...,1],cmap='gray')
fig.colorbar(fig_img,ax=axs[2],shrink=0.5)
axs[2].set_title('canal verde')

fig_img=axs[3].imshow(img_rgb[...,2],cmap='gray')
fig.colorbar(fig_img,ax=axs[3],shrink=0.5)
axs[3].set_title('canal azul')
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize']=[10,5]
size_imagen=100
R=np.ones((size_imagen,size_imagen))*255
G=np.ones((size_imagen,size_imagen))*128
B=np.ones((size_imagen,size_imagen))*0

imagen=np.dstack((R,G,B))
imagen=np.array(imagen,dtype=int)

plt.figure()
plt.imshow(imagen)
plt.title('imagen de color 1 naranja')
plt.show()
```

escala de grises y profundidad de color 

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('morado.jpg', cv2.IMREAD_GRAYSCALE)
original = img.dtype
print("Profundidad original:", original)

img16bits = np.uint16(img)
img_new = img16bits.dtype
print('Profundidad del color convertida:', img_new)

cv2.imwrite('morado_16bits.png', img16bits)


plt.figure(figsize=(10, 3)) 
plt.subplot(1,2,1)
plt.imshow(img16bits, cmap='gray')  
plt.title('Imagen original')


plt.subplot(1,2,2)
plt.title('imagen con 16bits')
plt.imshow(img16bits,cmap='gray')
```
