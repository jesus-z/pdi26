```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("chunli.jpg")

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

kernel1 = np.ones((6,6), np.float32) / 9

kernel2 = np.ones((10,10), np.float32) / 25

kernel3 = np.ones((12,12), np.float32) / 81

blur1 = cv2.filter2D(img, -1, kernel1)
blur2 = cv2.filter2D(img, -1, kernel2)
blur3 = cv2.filter2D(img, -1, kernel3)

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(blur1)
plt.title("Blur 3x3")
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(blur2)
plt.title("Blur 5x5")
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(blur3)
plt.title("Blur 9x9")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('dva.png', cv2.IMREAD_GRAYSCALE)
kernels = {
    "Enfoque": np.array([[0, -1, 0],[-1, 5, -1],[0, -1, 0]]),
    "Desenfoque": np.ones((3,3), np.float32) / 9,
    "Realce bordes": np.array([[0, 0, 0],[-1, 1, 0],[0, 0, 0]]),
    "Repujado": np.array([[-2, -1, 0],[-1,  1, 1],[ 0,  1, 2]]),
    "Detección bordes": np.array([[0,  1, 0],[1, -4, 1],[0,  1, 0]]),
    "Sobel": np.array([[-1, 0, 1],[-2, 0, 2],[-1, 0, 1]]),
    "Sharpen": np.array([[ 1, -2,  1], [-2,  5, -2], [ 1, -2,  1]]),
    "Norte": np.array([[ 1,  1,  1],[ 1, -2,  1],[-1, -1, -1]]),
    "Este": np.array([[-1,  1,  1],[-1, -2,  1],[-1,  1,  1]])
}
resultados = []
for nombre, kernel in kernels.items():
    filtrada = cv2.filter2D(img, -1, kernel)
    resultados.append((nombre, filtrada))

plt.figure(figsize=(20, 20))
plt.subplot(3, 4, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')
for i, (nombre, imagen) in enumerate(resultados):
    plt.subplot(3, 4, i + 2)
    plt.imshow(imagen, cmap='gray')
    plt.title(nombre)

plt.tight_layout()
plt.show()
```
