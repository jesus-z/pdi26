```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

def mostrar_imagen(imagen, titulo, es_gris=False):
    if es_gris:
        plt.imshow(imagen, cmap='gray')
    else:
    
        img_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
        plt.imshow(img_rgb)
    plt.title(titulo)
    plt.axis('off')


img1 = cv2.imread('yo.png') 

gris1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)

cv2.imwrite('foto1_bgr.png', img1)
cv2.imwrite('foto1_gris.png', gris1)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
mostrar_imagen(img1, 'Color Original (RGB)')

plt.subplot(1, 2, 2)
mostrar_imagen(gris1, 'Escala de Grises', es_gris=True)

plt.show()

b, g, r = cv2.split(img1)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(r, cmap='jet')
plt.title('Canal Rojo (Color Falso)')

plt.subplot(1, 3, 2)
plt.imshow(g, cmap='Greens')
plt.title('Canal Verde')

plt.subplot(1, 3, 3)
plt.imshow(b, cmap='Blues')
plt.title('Canal Azul')

plt.show()


img2 = cv2.imread('yo2.png') 

gris2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

cv2.imwrite('foto2_bgr.png', img2)
cv2.imwrite('foto2_gris.png', gris2)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
mostrar_imagen(img2, 'Color Original (RGB)')

plt.subplot(1, 2, 2)
mostrar_imagen(gris2, 'Escala de Grises', es_gris=True)

plt.show()

b, g, r = cv2.split(img2)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(r, cmap='jet')
plt.title('Canal Rojo (Color Falso)')

plt.subplot(1, 3, 2)
plt.imshow(g, cmap='Greens')
plt.title('Canal Verde')

plt.subplot(1, 3, 3)
plt.imshow(b, cmap='Blues')
plt.title('Canal Azul')

plt.show()


img3 = cv2.imread('yo3.png') 

gris3 = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)

cv2.imwrite('foto3_bgr.png', img3)
cv2.imwrite('foto3_gris.png', gris3)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
mostrar_imagen(img3, 'Color Original (RGB)')

plt.subplot(1, 2, 2)
mostrar_imagen(gris3, 'Escala de Grises', es_gris=True)

plt.show()

b, g, r = cv2.split(img3)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(r, cmap='jet')
plt.title('Canal Rojo (Color Falso)')

plt.subplot(1, 3, 2)
plt.imshow(g, cmap='Greens')
plt.title('Canal Verde')

plt.subplot(1, 3, 3)
plt.imshow(b, cmap='Blues')
plt.title('Canal Azul')

plt.show()
```
