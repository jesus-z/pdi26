```python
import cv2

import numpy as np

from matplotlib import pyplot as plt


img = cv2.imread('hul.png', cv2.IMREAD_GRAYSCALE)


laplacian_simple = cv2.Laplacian(img, cv2.CV_64F)


blur = cv2.GaussianBlur(img, (9, 9), 0)

log = cv2.Laplacian(blur, cv2.CV_64F)



plt.figure(figsize=(35, 25))



plt.subplot(1, 3, 1)

plt.imshow(img, cmap='gray')

plt.title('Original / Gris')

plt.axis('off')



plt.subplot(1, 3, 2)

plt.imshow(laplacian_simple, cmap='gray')

plt.title('Laplaciano Simple\n(Sensible al ruido)')

plt.axis('off')



plt.subplot(1, 3, 3)

plt.imshow(log, cmap='gray')

plt.title('Laplaciano de Gaussiano (LoG)\n(Bordes limpios)')

plt.axis('off')



plt.tight_layout()
```
